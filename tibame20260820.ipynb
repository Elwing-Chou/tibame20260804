```python
f = open("a.txt", "w", encoding="utf-8")
# 以前的(function): write(f, "abcde")
# 專屬某種型態的功能(method): f.write("abcde")
f.write("{ab}cde")
# \字: \n \t \b
f.write("\nghi")
# 以前: close(f)
# 專屬功能: f.close()
f.close()
```

```python
f = open("a.txt", "r", encoding="utf-8")
# content = read(f)
content = f.read()
print(content)
f.close()
```

- 字典: 聚集來表示一個複雜單位

```python
l = [20, 30, 40]

# !
print(l[0])
print(max(l), min(l), sum(l), len(l))
print(l[len(l)-1])
# !
print(l[-1])

l[0] = 99
print(l)

for n in l:
    print("!", n)

for i in range(len(l)):
    print("!", l[i])
```

- JSON格式: [], {}
- 標準表格格式(CSV=comma separated values)
    - 姓名,身高
    - Elwing,175
    - "Bob, Mr.Chou",180

```python
import json
import urllib.request as req
import pandas as pd

url = "https://api.gamer.com.tw/anime/v1/danmu.php?videoSn=38250&geo=TW%2CHK"
f = req.urlopen(url)
content = f.read()
# print(type(content))
# 型態轉換: bytes-> []/{}
content = json.loads(content)
# print(type(content))

danmus = content["data"]["danmu"]
# for d in danmus:
#     print(d["userid"], d["text"])

# list/dict -> pandas自己定義的 DataFrame型態
df = pd.DataFrame(danmus)
# index=False: 不存列標籤
df.to_csv("baha.csv", encoding="utf-8", index=False)
df.to_excel("baha.xlsx")
df
```
