```python
f = open("a.txt", "w", encoding="utf-8")
# 以前的(function): write(f, "abcde")
# 專屬某種型態的功能(method): f.write("abcde")
f.write("abcde")
# \字: \n \t \b
f.write("\nghi")
# 以前: close(f)
# 專屬功能: f.close()
f.close()
```

```python
f = open("a.txt", "r", encoding="utf-8")
# content = read(f)
content = f.read()
print(content)
f.close()
```
