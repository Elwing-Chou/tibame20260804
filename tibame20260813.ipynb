```python
# ****
#  ***
#   **
#    *
# 空白建: 0 1 2 3
# *: 4 3 2 1
# i: 0 1 2 3
symbol1, symbol2 = " ", "o"
layer = 4
i = 0
while i < layer:
    count1 = i
    count2 = layer - i
    line = symbol1 * count1 + symbol2 * count2
    print(line)
    i = i + 1
```

```python
# ooooooo
#  ooooo
#   ooo
#    o
# 空白建: 0 1 2 3
# o: 7 5 3 1
# i: 0 1 2 3
# -2*i: 0 -2 -4 -6

# layer 第一層
# 4 7
# 5 9
# 6 11
# layer * 2 -> 8 10 12

layer = 5
s1, s2, s3 = " ", "o", " "
i = 0
while i < layer:
    c1 = i
    c2 = (2 * layer - 1) - 2 * i
    c3 = i
    line = s1 * c1 + s2 * c2 + s3 * c3
    print(line)
    i = i + 1
```

```python
# 我們想畫出sin的圖形
# 我們的 現實坐標系(往上增加) -> 程式座標(往下增加的)
# 所以你原本算出來的y -> 加負號 -y
# 我們想把它移到中間: 算出來是0的時候 + 250(500/2)

# x軸: 把它當成角度(x=180->sin180度)

# y不做一個放大, 你的值就是 -1 1->你根本看不到 sin 250*sin
import math
from ColabTurtle.Turtle import *
initializeTurtle(initial_speed=2, initial_window_size=(800,500))

# 調整初始位置
penup()
goto(0, 250)
pendown()

# x: angle
angle = 90
radian = math.radians(angle)
y = 250 * math.sin(radian)
y_cor = -1 * y + 250
goto(angle, y_cor)

angle = 180
radian = math.radians(angle)
y =  250 * math.sin(radian)
y_cor = -1 * y + 250
goto(angle, y_cor)

angle = 270
radian = math.radians(angle)
y =  250 * math.sin(radian)
y_cor = -1 * y + 250
goto(angle, y_cor)

angle = 360
radian = math.radians(angle)
y =  250 * math.sin(radian)
y_cor = -1 * y + 250
goto(angle, y_cor)
```

```python
# 把0-800分成n等分: 100分 0 8度 16度 24度 .... 做100次
import math
from ColabTurtle.Turtle import *

width, height = 720, 500
initializeTurtle(initial_speed=10, initial_window_size=(width, height))

# 調整初始位置
height_half = height / 2
penup()
goto(0, height_half)
pendown()

n = 200
interval = width / n
# -1~1太小了, 我乘上一個放大 250 sin
ratio = 100

# angle = 0  630
# i = 0....7
i = 0
while i < n + 1:
    # x: angle
    # 800 / 100 -> 8度
    # 度 0 8 16 24
    # i  0 1 2 3
    angle = i * interval
    radian = math.radians(angle)
    y = ratio * math.sin(radian)
    y_cor = -1 * y + height_half
    goto(angle, y_cor)
    i = i + 1
```
