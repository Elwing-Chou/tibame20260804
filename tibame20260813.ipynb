```python
# ****
#  ***
#   **
#    *
# 空白建: 0 1 2 3
# *: 4 3 2 1
# i: 0 1 2 3
symbol1, symbol2 = " ", "o"
layer = 4
i = 0
while i < layer:
    count1 = i
    count2 = layer - i
    line = symbol1 * count1 + symbol2 * count2
    print(line)
    i = i + 1
```
