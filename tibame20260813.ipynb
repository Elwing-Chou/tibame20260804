```python
# *
# **
# ***
# ****
i = 0
while i < 5:
    print("o" * (i + 1))
    i = i + 1
```

```python
# ****
#  ***
#   **
#    *
# 空白建: 0 1 2 3
# *: 4 3 2 1
# i: 0 1 2 3
symbol1, symbol2 = " ", "o"
layer = 4
i = 0
while i < layer:
    count1 = i
    count2 = layer - i
    line = symbol1 * count1 + symbol2 * count2
    print(line)
    i = i + 1
```

```python
# ooooooo
#  ooooo
#   ooo
#    o
# 空白建: 0 1 2 3
# o: 7 5 3 1
# i: 0 1 2 3
# -2*i: 0 -2 -4 -6

# layer 第一層
# 4 7
# 5 9
# 6 11
# layer * 2 -> 8 10 12

layer = 5
s1, s2, s3 = " ", "o", " "
i = 0
while i < layer:
    c1 = i
    c2 = (2 * layer - 1) - 2 * i
    c3 = i
    line = s1 * c1 + s2 * c2 + s3 * c3
    print(line)
    i = i + 1
```
