```python
import urllib.request as req
import bs4 as bs

def download_page(page):
    url = f"https://tabelog.com/tw/tokyo/rstLst/sweets/{page}/?SrtT=rt"
    resp = req.urlopen(url)
    content = resp.read()
    html = bs.BeautifulSoup(content)
    rs = html.find_all("div", {"class":"list-rst__body"})
    # !!!
    result = []
    for r in rs:
        # 評價/晚價/午價/例假日
        name = r.find("a", {"class":"list-rst__rst-name-target"})
        area_genre = r.find("div", {"class":"list-rst__area-genre"})
        rating = r.find("span", {"class":"c-rating__val"})
        prices = r.find_all("span", {"class":"c-rating-v3__val"})
        lunch_price = prices[1]
        dinner_price = prices[0]
        holiday = r.find("span", {"class":"list-rst__holiday-text"})

        name_text = name.get_text().strip()
        name_href = name["href"]
        area_genre_text = area_genre.get_text().strip()
        rating_text = rating.get_text().strip()
        lunch_price_text = lunch_price.get_text().strip()
        dinner_price_text = dinner_price.get_text().strip()
        holiday_text = holiday.get_text().strip()
        # !
        data = {
            "rating":rating_text,
            "name":name_text,
            "link":name_href,
            "area_genre":area_genre_text,
            "price_lunch":lunch_price_text,
            "price_dinner":dinner_price_text,
            "holiday":holiday_text,
        }
        result.append(data)
    return result

download_page(2)
```

- 迴圈SOP

```python
import pandas as pd
total = []
for i in range(10):
    page = i + 1
    result_json = download_page(page)
    result_df = pd.DataFrame(result_json)
    total.append(result_df)

df = pd.concat(total)
# if you need
df = df.reset_index(drop=True)
df
```

```python
def func(i):
    return i * 3

df["rating"].astype(float).apply(func)
```

```python
def func(i):
    spl = i.split(" / ")
    ans = {
        "area":None,
        "genre":None
    }
    if len(spl) == 2:
        ans["area"] = spl[0]
        ans["genre"] = spl[1]
    elif len(spl) == 1:
        ans["genre"] = spl[1]
    return pd.Series(ans)

df[["area", "genre"]] = df["area_genre"].apply(func)
df
```

```python
tags = [
    "星期一",
    "星期二",
    "星期三",
    "星期四",
    "星期五",
    "星期六",
    "星期日"
]
def func(i):
    ans = {}
    for t in tags:
        if t in i:
            ans[t] = "N"
        else:
            ans[t] = "Y"
    return pd.Series(ans)


df[tags] = df["holiday"].apply(func)
fil = df["星期一"] == "Y"
df[fil]
```
