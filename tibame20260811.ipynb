- 清單(list): ["剪刀", "時投"]
    - python會偷偷給清單的每一個位置一個編號(key): 0 1 2 ....
    - 你只要有key/value: 查詢操作 清單名字[key]

```python
import random

p1 = int(input("0.剪刀 1.石頭 2.布:"))
p2 = random.randint(0, 2)

tags = ["剪刀", "石頭", "布"]

print("我選的:", tags[p1])
print("他選的:", tags[p2])
```
