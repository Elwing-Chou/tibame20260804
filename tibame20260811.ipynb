- 7 / 3 = 2...1
    - 取商: 7 // 3 -> 2      分組操作
    - 取餘(mod): 7 % 3 -> 1  循環操作

````python
import random

p1 = int(input("0.剪刀 1.石頭 2.布:"))
p2 = random.randint(0, 2)

tags = ["剪刀", "石頭", "布"]

print("我選的:", tags[p1])
print("他選的:", tags[p2])

# 我選的是電腦選的下一個
if p1 == (p2 + 1) % 3:
    print("win")
elif p1 == (p2 - 1) % 3:
    print("lose")
else:
    print("even")
````

````python
import random

# p1 = int(input("0-5:"))
# p2 = random.randint(0, 5)
p1 = 3
p2 = 5

tags = ["剪刀(0)", "石頭(0)", "布(0)", "剪刀(1)", "石頭(1)", "布(1)"]

print("我選的:", tags[p1])
print("他選的:", tags[p2])

# 確認是否同一組
p1_group = p1 // 3
p2_group = p2 // 3
if p1_group == p2_group:
    print("同組")
    # case1: 012
    # case2: 345 -> 012
    p1_new = p1 % 3
    p2_new = p2 % 3
    if p1_new == (p2_new + 1) % 3:
        print("win")
    elif p2_new == (p1_new + 1) % 3:
        print("lose")
    else:
        print("even")
else:
    print("不同組")
````

- 迴圈(重複)
    - 重複固定次數的迴圈, 需要一個計數器, 從0屬到9(10次)
        - 開始: i = 0
        - 增加: i = i + 1
        - 判斷: i < 10

- 養成好習慣: 寫程式要無腦(找到一個好的做法, 接下來永遠這麼做)
    - 上面三條千萬別給我換
        - 次數在<後面
        - i = 0...次-1

````python
i = 0
while i < 5:
    print(i, 4-i, 2*i+1)
    i = i + 1
````
