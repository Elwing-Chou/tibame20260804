```python
import json
import urllib.request as req
import pandas as pd
import os

# 如果資料夾不存在, 就創造資料夾
def download_danmu(sn):
    dirname = "baha/鬼滅"
    if not os.path.exists(dirname):
        os.makedirs(dirname)

    # sn = "38250"
    url = f"https://api.gamer.com.tw/anime/v1/danmu.php?videoSn={sn}&geo=TW%2CHK"
    f = req.urlopen(url)
    content = f.read()
    content = json.loads(content)

    danmus = content["data"]["danmu"]

    df = pd.DataFrame(danmus)
    df.to_csv(f"{dirname}/{sn}.csv", encoding="utf-8", index=False)
    return df

download_danmu("38331")
```

- HTTP代碼
    - 2: 成功
    - 3: 轉址
    - 4: 錯誤
        - 404: Not Found
            - 通常都是你網址打錯
        - 403: Forbidden
            - ip 被 ban
                - 後面幾次才被拒絕
                - 最好不要送網址送太快
                - 如果真的被 ban -> 等/換ip
            - Headers不夠完整
                - 最常被檢查的是user-agent

- 問題和回復
    - 網址+額外訊息(request headers) !!! 控制
    - 回復+額外訊息(response headers)

```python
import urllib.request as req

# url = "https://ani.gamer.com.tw/animeVideo.php?sn=38250"
url = "https://github.com/Elwing-Chou/tibame20260804/raw/refs/heads/main/baha.html"
h = {
    "User-Agent":"Mozilla/5.0"
}
r = req.Request(url, headers=h)
resp = req.urlopen(r)
content = resp.read()
print(content)
```
