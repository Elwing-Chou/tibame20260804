```python
# range(5) -> [0, 1, 2, 3, 4] !!!!!!!!
# range(2, 6) -> [2, 3, 4, 5]
# range(1, 9, 2) -> [1, 3, 5, 7]
for i in range(10):
    n = 10 - i
    print(n)
```

```python
# 如果你在處理的時候腦袋有個東西, 準備一組名字=值
ans = 0
for i in range(10):
    ans = ans + (i + 1)
    print(ans)
```
