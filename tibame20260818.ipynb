```python
# 剪刀石頭布: 先贏三把得算贏
# while 自由寫
# 我不重寫剪刀石頭不, 我隨機 randint(-1, 1) -1: lose 0: even 1: win
import random

win, even, lose = 0, 0, 0
tags = ["輸", "平手", "贏"]
while (win < 3) and (lose < 3):
    result = random.randint(-1, 1)
    print("結果:", tags[result+1])
    if result == -1:
        lose = lose + 1
    elif result == 1:
        win = win + 1
    elif result == 0:
        even = even + 1

print(win, even, lose)
```
