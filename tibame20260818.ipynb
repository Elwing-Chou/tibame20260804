- part3. 迴圈: 不是用次數來規範的迴圈
- 布林型態(是否): True False
    - 不: not a > 3
    - 組合多個條件
        - and: 而且
        - or: 或者

```python
# 剪刀石頭布: 先贏三把得算贏
# while 自由寫
# 我不重寫剪刀石頭不, 我隨機 randint(-1, 1) -1: lose 0: even 1: win
import random

win, even, lose = 0, 0, 0
tags = ["輸", "平手", "贏"]
while (win < 3) and (lose < 3):
    result = random.randint(-1, 1)
    print("結果:", tags[result+1])
    if result == -1:
        lose = lose + 1
    elif result == 1:
        win = win + 1
    elif result == 0:
        even = even + 1

print(win, even, lose)
```

```python
# 無腦: 只要不是用次數規範的迴圈, while 都直接 pass -> while True(無窮迴圈) + break
# 優點
# 1. 時間很自由
# 2. 收尾工作很自由
# 3. 組合也很自由
import random

win, even, lose = 0, 0, 0
tags = ["輸", "平手", "贏"]
while True:
    result = random.randint(-1, 1)
    print("結果:", tags[result+1])
    if result == -1:
        lose = lose + 1
    elif result == 1:
        win = win + 1
    elif result == 0:
        even = even + 1
    # judge
    if win == 3:
        break
    elif lose == 3:
        break

print(win, even, lose)
```

```python
# DEUCE:
# 贏的分數是5, 但是如果 5:4 時候 沒有結束 6:4
# 如果超過門檻後(對方只差你一分以內), 就要贏兩分才結束
import random

threshold = 5
win, lose = 0, 0
tags = ["輸", "贏"]
while True:
    # 0: lose 1: win
    result = random.randint(0, 1)
    print(tags[result])
    if result == 0:
        lose = lose + 1
    else:
        win = win + 1
    # 開始要判斷輸贏
    if win >= threshold or lose >= threshold:
        # 如果對手和我差一分以內
        if abs(win - lose) <= 1:
            pass
        # 兩分以上
        else:
            print(win, lose, "比賽結束")
            break
    # 根本不用判斷輸贏
    else:
        pass
```

```python
import random

threshold = 5
win, lose = 0, 0
deuce = 0
tags = ["輸", "贏"]
while True:
    # 0: lose 1: win
    result = random.randint(0, 1)
    if result == 0:
        lose = lose + 1
    else:
        win = win + 1
    print(tags[result], win, "-", lose)

    # 分數差兩分以上
    if abs(win - lose) >= 2:
        if win >= threshold or lose >= threshold:
            print(win, lose, "比賽結束")
            break
        else:
            pass
    # 沒有
    else:
        pass

    # 如果比賽還沒結束, 但是有人到達分數線=deuce
    if win >= threshold or lose >= threshold:
        deuce = deuce + 1

print("Deuce:", deuce)
```
