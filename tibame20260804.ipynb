```python
# 註解
# bmi程式:
# 1. 可以讓使用者輸入自己的身高體重
# 2. 根據身高體重計算出使用者的 bmi=體重(kg)/身高(m)^2
# 3. 根據這個計算出的bmi來給使用者建議

weight = 77
height = 175
bmi = weight / (height / 100) ** 2
print("體重(kg):" + str(weight))
print("身高(cm):" + str(height))
print("bmi是:" + str(bmi))
```
