- input()跟print()有點不太一樣
    - print是列印, 沒有回覆
    - input是有回覆(答案)的, 就是使用者輸入的那個東西
    - 這個概念我們叫做回傳值(答案)
    - 參數/回傳值
    - input回傳(新)值: 確認型態!!!!!!!!!!!!!!!!

```python
# 註解
# bmi程式:
# 1. 可以讓使用者輸入自己的身高體重
# 2. 根據身高體重計算出使用者的 bmi=體重(kg)/身高(m)^2
# 3. 根據這個計算出的bmi來給使用者建議
weight = float(input("請輸入您的體重(kg):"))
height = float(input("請輸入您的身高(cm):"))
bmi = weight / (height / 100) ** 2
print("體重(kg):" + str(weight))
print("身高(cm):" + str(height))
print("bmi是:" + str(bmi))
```

```python
# 科普
print(3+2.0)
# 兩個小數相等
# 很多人覺得每個數字都要一模一樣才叫做兩小數相等(X)
# 問題: 我把一個pizza(1.0)分成三塊(0.333333333) -> 0.3333333333 * 3 ?= 1.0(相等)
# 因為小數是不精準數(無窮/進位轉換有限無限), 所以正確的小數比較: 兩個小數相減是一個很小很小的差距
# 浮點數誤差(正常: 本來計算/轉換進位就會有誤差)
print(3+3.14)
```

```python
# 回傳值: 做完這個功能所回覆的一個"新的"答案(值)
# 正確的講法 不是 把4.2轉成4
#             是 利用傳進來的4.2 做出一個 4(答案) 給你
# 守則: 你只要不做=, 值就永遠不會改
a = 4.2
b = int(a)
print(a)
print(b)
```
