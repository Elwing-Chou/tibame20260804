```python
# no module named distutils: pip install --upgrade setuptools
import undetected_chromedriver as uc
import time
import re
import os
import urllib.request as req
import urllib.parse as parse
# MAC: 不加會出現SSL Certificate failed
import ssl
ssl._create_default_https_context = ssl._create_unverified_context
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

if __name__ == "__main__":

    name = "google"
    if not os.path.exists(name):
        os.makedirs(name)

    driver = uc.Chrome(version_main=152, use_subprocess=False)
    driver.get('https://www.google.com/')
    driver.maximize_window()
    # bs: find find_all
    # selenium: find_element find_elements
    e = driver.find_element(By.CLASS_NAME, "gLFyf")
    # 點: click() 輸入: send_keys()
    e.send_keys("吉伊卡哇")
    e.send_keys(Keys.ENTER)
    time.sleep(2)
    e = driver.find_elements(By.CLASS_NAME, "C6AK7c")[2]
    e.click()
    time.sleep(3)
    es =  driver.find_elements(By.TAG_NAME, "img")
    for e in es:
        try:
            # bs: .get_text() ["href"]
            # selenium: .text(專屬值) .get_attribute("href")
            # 每種型態可以擁有兩種東西: 專屬功能 專屬值
            url = e.get_attribute("src")
            print(url)
            h = {
                "User-Agent": "Mozilla/5.0"
            }
            r = req.Request(url, headers=h)
            resp = req.urlopen(r)
            content = resp.read()

            f = open(f"{name}/{time.time()}.png", "wb")
            f.write(content)
            f.close()
        except:
            print(url, "發生錯誤")

    time.sleep(5)

# if else: 事前預防
# try except: 事後解決
# try:
#   發生錯誤
# except XXError:
#   解決方案
# 1. 順利執行: 不會跑except
# 2. 發生錯誤(中斷接下來的try):
# a. 你有except: 做except
# b. 你沒有except: 跳出紅字
```
