- 前端(local)
    - 呈現網頁
        - HTML: 網頁內容
        - CSS: 網頁排版
            - !!!! 你要找的區塊通常會用特定排版, 透過排版來篩選到他 !!!!
        - JavaScript: 網頁動態
            - 很多時候控制你隱藏網址的觸發
            - 不重要, 因為99%的時候你根本不care誰送的

- 後端(server)
    - 收網址(http server跑在某個port去接收)
    - 處理這個網址(python Flask/Django/fastAPI)
    - 處理網址: 查資料庫(SQL)
    - 送回回覆(html/mp3/mp4...)

- CSS 簡介: 找到我想做排版的區塊, 加上排版
    - 會特別加上可以被分辨的HTML屬性, 來讓他跟別人不一樣
    - class="分類1 分類2 分類3..."
    - id="身分證"

- CSS語法:
    - 利用html名稱:
        - a{background-color:red}
    - 利用class名稱:
        - .js-thumbnail-img{border-radius:50%}
    - 利用id:
        - #id名稱{}

```python
import urllib.request as req
import bs4 as bs

url = "https://tabelog.com/tw/tokyo/rstLst/sweets/?SrtT=rt"
resp = req.urlopen(url)
content = resp.read()
html = bs.BeautifulSoup(content)
rs = html.find_all("div", {"class":"list-rst__body"})
for r in rs:
    name = r.find("a", {"class":"list-rst__rst-name-target"})

    name_text = name.get_text()
    name_href = name["href"]

    print(name_text)
    print(name_href)
    print("-" * 30)
```
